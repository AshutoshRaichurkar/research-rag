```python
from research_rag.evals.judge import judge_faithfulness, judge_relevance
from research_rag.rag import answer

result = answer("How does RAGAS calculate the faithfulness score?")
print(result.text, "\n")

faithfulness = judge_faithfulness(result.text, result.chunks)
if faithfulness is None:
    print("judge_error")
else:
    for claim in faithfulness.claims:
        mark = "✅" if claim.supported else "❌"
        print(mark, claim.claim)
    print("faithfulness:", faithfulness.score())

relevance = judge_relevance(result.question, result.text)
print("relevance:", relevance)
```
